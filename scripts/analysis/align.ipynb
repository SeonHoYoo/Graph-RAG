# Align Analysis

Input 파일에서 `answer_matches_gold` 와 `align` 두 열의 분포를 테이블로 확인합니다.

```python
from pathlib import Path
import json
import pandas as pd

def align_table(input_path):
    with input_path.open("r", encoding="utf-8") as f:
        records = json.load(f)

    df = pd.DataFrame(records)
    df[["answer_matches_gold", "align"]].head()
    answer_group = df["answer_matches_gold"].map(lambda x: "true" if x is True else "false")

    table = pd.crosstab(
        answer_group,
        df["align"].fillna("null").astype(str),
        dropna=False,
    )
    table.index.name = "answer"
    table.columns.name = "align"

    table_with_total = table.copy()
    table_with_total["total"] = table_with_total.sum(axis=1)
    table_with_total.loc["total"] = table_with_total.sum(axis=0)
    return table_with_total
```
