# Matches Analysis

Input 파일에서 `answer_matches_gold` 와 `document_supported` 두 열의 분포를 테이블로 확인합니다.

```python
from pathlib import Path
import json
import pandas as pd

def matches_table(input_path):
    with input_path.open("r", encoding="utf-8") as f:
        records = json.load(f)

    df = pd.DataFrame(records)
    df[["answer_matches_gold", "document_supported"]].head()

    answer_group = df["answer_matches_gold"].map(
        lambda x: "true" if x is True else "false"
    )
    support_group = df["document_supported"].fillna("null").astype(str)

    table = pd.crosstab(
        answer_group,
        support_group,
        dropna=False,
    )
    table.index.name = "answer"
    table.columns.name = "supported"

    table_with_total = table.copy()
    table_with_total["total"] = table_with_total.sum(axis=1)
    table_with_total.loc["total"] = table_with_total.sum(axis=0)
    return table_with_total
```
